```python
# Task 4: Describe the numbers
num_cols = ['bedrooms', 'beds', 'availability_365', 'minimum_nights']
listings[num_cols].describe().round(2)

# The column with the biggest maximum is 'minimum_nights' with a max value of 500 nights.
# A minimum stay of 500 nights (around 1.4 years) does not seem realistic for standard short-term holiday rentals.
```

```python
# Task 5: Describe the text
print(listings[['property_type', 'neighbourhood_cleansed']].describe())
print("\nUnique count for property_type:", listings['property_type'].nunique())

# 'property_type' has 44 unique values. The most common is 'Entire rental unit' (2442 times)
# 'neighbourhood_cleansed' has 14 unique values.The most common is 'Södermalms' (1145 times).
```

```python
# Task 6
print("First 3 prices:", listings['price'].head(3))
listings[['name', 'price']].sort_values('price', ascending=False).head(3)

# The 'price' column is stored as text and is sorted alphabetically (lexicographically) 
```

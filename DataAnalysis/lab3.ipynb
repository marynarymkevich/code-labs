```python
import pandas as pd

loans_raw = pd.read_csv('data/loans.csv')
loans = loans_raw.copy()

print(loans.shape)
loans.info()

# Problems: 
# loan_date, return_date, late_fee and returned are str, but should be dates, numbers and booleans
# notes is completely empty
# return_date has missing values (593 out of 623)
```

Task 2

```python
print(loans['genre'].nunique())
print(loans['genre'].str.strip().str.lower().nunique())
print(loans['genre'].str.strip().str.lower().unique())

loans['genre'] = (
    loans['genre']
    .str.strip()
    .str.lower()
    .replace({
        "children's": 'children',
        'science fiction': 'sci-fi',
        'sci fi': 'sci-fi'
    })
    .str.title()
)

print(loans['genre'].nunique())
loans['genre'].value_counts()
```

Task 3

```python
loans['returned'].nunique()
loans['returned'].unique()
loans['returned'] = (
    loans['returned']
    .str.strip()
    .str.lower()
    .map({
        'yes': True,
        'y': True,
        'no': False,
        'n': False
    })
)
print(loans['returned'].isna().sum())
print(loans['returned'].nunique())
print(loans['returned'].value_counts())
```

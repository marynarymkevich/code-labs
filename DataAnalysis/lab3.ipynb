Task 1

```python
import pandas as pd

loans_raw = pd.read_csv('data/loans.csv')
loans = loans_raw.copy()

print(loans.shape)
loans.info()

# Problems: 
# loan_date, return_date, late_fee and returned are str, but should be dates, numbers and booleans
# notes is completely empty
# return_date has missing values (593 out of 623)
```

Task 2

```python
print(loans['genre'].nunique())
print(loans['genre'].str.strip().str.lower().nunique())
print(loans['genre'].str.strip().str.lower().unique())

loans['genre'] = (
    loans['genre']
    .str.strip()
    .str.lower()
    .replace({
        "children's": 'children',
        'science fiction': 'sci-fi',
        'sci fi': 'sci-fi'
    })
    .str.title()
)

print(loans['genre'].nunique())
loans['genre'].value_counts()
```

Task 3

```python
loans['returned'].nunique()
loans['returned'].unique()
loans['returned'] = (
    loans['returned']
    .str.strip()
    .str.lower()
    .map({
        'yes': True,
        'y': True,
        'no': False,
        'n': False
    })
)
print(loans['returned'].isna().sum())
print(loans['returned'].nunique())
print(loans['returned'].value_counts())
```

Task 4

```python
# loans['late_fee'].astype(float)

fee = (
    loans['late_fee']
    .str.strip()
    .str.replace('kr', '')
    .str.replace(' ', '')
    .str.replace(',', '.')
 )


loans['late_fee'] = fee.astype(float)
print(loans['late_fee'].describe().round(2))
print(loans['late_fee'].sum())
```

Task 5

```python
dates = pd.to_datetime(loans['loan_date'], errors='coerce')
print(dates.isna().sum())

loans['loan_date'] = dates
print(loans['loan_date'].min())
print(loans['loan_date'].max())

print(loans['loan_date'].dt.to_period('M').value_counts().sort_index())
print("Monthly sum:", loans['loan_date'].dt.to_period('M').value_counts().sum())
print("Missing dates:", loans['loan_date'].isna().sum())
```

Task 6

```python
print("Duplicates in raw table:", loans_raw.duplicated().sum())
print("Duplicates after cleaning:", loans.duplicated().sum())
print("Duplicates by loan_id:", loans.duplicated(subset='loan_id').sum())

# In raw table some duplicate rows had small formatting differences (e.g. 'ROMANCE' vs 'romance')
# After cleaning these rows became the same

total_with_duplicates = loans['late_fee'].sum()
print("Total late fees with duplicates:", total_with_duplicates)

loans = loans.drop_duplicates().reset_index(drop=True)

total_without_duplicates = loans['late_fee'].sum()
print("Total late fees without duplicates:", total_without_duplicates)

diff = total_with_duplicates - total_without_duplicates
diff_pct = (diff / total_with_duplicates) * 100
print("Difference in kr:", diff)
print("Percentage difference:", diff_pct, "%")
```

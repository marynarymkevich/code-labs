# Go Deeper
# Task 1

```python
import pandas as pd
import numpy as np

listings = pd.read_csv('data/listings.csv')
print(listings.shape)

listings['no_reviews'] = listings['number_of_reviews'] == 0
total_unreviewed = listings['no_reviews'].sum()
share_unreviewed = listings['no_reviews'].mean()
print(f"Total unreviewed: {total_unreviewed} listings ({share_unreviewed:.3f})")

unreviewed_by_neighbourhood = (
    listings.groupby('neighbourhood_cleansed').agg(
        number_of_listings=('id', 'count'),
        share_no_reviews=('no_reviews', 'mean')
    )
    .sort_values('share_no_reviews')
    .round(3)
)
print(unreviewed_by_neighbourhood)
```

# Task 2

```python
reviewed = listings[listings['number_of_reviews'] > 0]
print(reviewed.shape)

review_labels = ['few', 'some', 'many', 'most']
reviewed['review_group'] = pd.qcut(
    reviewed['number_of_reviews'],
    q=4,
    labels=review_labels
)

print(f"Counts per review group: {reviewed['review_group'].value_counts()}")
# They are not exactly equal because listings with the same number of reviews cannot be divided and go to the same group


print(pd.qcut(reviewed['number_of_reviews'], 4).cat.categories)
# The borders of the four groups are: 1, 4, 13, 36, and 1293.
```

# Task 3

```python
(
    reviewed.groupby('review_group')['review_scores_rating']
    .mean()
    .round(2)
)

# What we can conclude:
# There is no clear positive relationship between review volume and rating
# Ratings stay remarkably consistent and high (~4.8) across all four groups

# What we cannot conclude:
# We cannot conclude that more reviews improves or worsens listing quality.
# We cannot say anything about unreviewed listings because they were excluded from our analysis
```

# Task 4

```python
(
    reviewed.groupby(['neighbourhood_cleansed', 'room_type'])
    .agg(
        number_of_listings=('id', 'count'),
        avg_accommodates=('accommodates', 'mean'),
    )
    .query('number_of_listings >= 30')
    .sort_values('avg_accommodates', ascending=False).round(2)
 )

# The second list is more trustworthy because it filters groups that have fewer than 30 listings,
# ensuring that the averages are calculated from statistically reliable sizes
```
